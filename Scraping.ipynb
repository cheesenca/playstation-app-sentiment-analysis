```python
import pandas as pd
pd.options.mode.chained_assignment = None

import re
import csv
import nltk
import string
import matplotlib.pyplot as plt
from langdetect import detect, DetectorFactory
from google_play_scraper import reviews_all, Sort

nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# Scrapping "PlayStation App" Reviews From Play Store
DetectorFactory.seed = 0
scrapreview = reviews_all(
    'com.scee.psxandroid',
    lang='en',
    country='us',
    sort=Sort.MOST_RELEVANT,
    count=100000
)
```

```python
# Load The Dataset
df = pd.DataFrame(scrapreview)
review_total, column_total = df.shape
```

```python
df = df.drop(columns=['replyContent', 'repliedAt'])
```

```python
df.head()
```

```python
# Remove Missing Value
clean_df = df.dropna()
clean_df = clean_df.drop_duplicates()
review_duplicate_remove, column_duplicate_remove = clean_df.shape
```

```python
clean_df.info()
```

```python
# Save Reviews in CSV File
with open("application_review.csv", mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['review'])
    for review in scrapreview:
        writer.writerow([review['content']])
```
